# Resumen: Clase 7 - Cadenas, Entradas y Salidas en Python


## 1. Entrada y Salida Estándar

- *Salida Estándar (stdout):** Por defecto, es la terminal o pantalla. En Python se gestiona con la función print().
- Sintaxis: print(expresión1, expresión2, ..., sep=' ', end='\n')
- *Nota sobre el tipo de retorno:** La función print() realiza una acción (imprimir) pero no retorna ningún valor, por lo que su ejecución devuelve `None (tipo NoneType).

```python
# Ejemplo 1: Multiplicación de strings (repetición)
print(3 * "Hola")

# Ejemplo 2: Imprimir variables y concatenar tipos de datos
a = 2
print("El valor de a es:", a)

# Ejemplo 3: El tipo de retorno de la función print()
print("Tipo de print(a):", type(print(a)))

# Ejemplo 4: Concatenación manual vs paso de argumentos múltiples
dinero = 30
print("Hoy gané $" + str(dinero), "en mi trabajo")
```

## 2. El Método splitlines()

Este método analiza una cadena de caracteres y la divide en una lista de subcadenas basándose en los caracteres de escape de salto de línea (\n, \r).

*Sintaxis:*<objeto str>.splitlines ()

```python
# Ejemplo 1: Dividir un string con nombres separados por salto de línea
nombres = "Hugo.\nPaco.\nLuis."
print("Texto original:")
print(nombres)
print("\nResultado de splitlines():")
print(nombres.splitlines())

# Ejemplo 2: Fragmentar una estrofa
himno = "Mexicanos al grito de guerra\nel acero aprestad y el bridón"
print(himno.splitlines())
```

## 3. El Método index()

Busca una subcadena dentro de un texto de izquierda a derecha y devuelve el índice de la primera coincidencia exacta.

- *Sintaxis:* <objeto str>.index(<subcadena>)
- *Importante:* Si la subcadena no existe, Python arrojará un error de tipo ValueError.

```python
texto = 'Luis, Hugo, Paco, Luis'

# Ejemplo 1: Buscar un término existente en medio de la cadena
print("Índice de 'Paco':", texto.index('Paco'))

# Ejemplo 2: Buscar un término repetido (devuelve la primera coincidencia)
print("Índice de 'Luis':", texto.index('Luis'))

# Ejemplo 3: Buscar un término inexistente (Generará un error ValueError)
try:
    texto.index('Juan')
except ValueError as e:
    print("Error capturado correctamente:", e)
```

## 4. El Método replace



Reemplaza todas las apariciones de una subcadena de búsqueda por una nueva cadena de sustitución. Devuelve una nueva cadena de texto sin modificar la original.

- *Sintaxis: <objeto str>.replace(<búsqueda>, <sustitución>)
- *Nota: Si no se encuentra el término de búsqueda, devuelve la cadena original sin cambios.

```python
cadena_original = 'Luis, Hugo, Paco, Luis'

# Ejemplo 1: Reemplazar un nombre por otro
nueva_cadena = cadena_original.replace('Luis', 'Juan Antonio')
print("Resultado del reemplazo:", nueva_cadena)

# Ejemplo 2: Intentar reemplazar un término inexistente (no cambia nada)
sin_cambios = cadena_original.replace('Pedro', 'Juan')
print("Sin cambios:", sin_cambios)

# Ejemplo 3: Comparación avanzada de cambios usando listas compuestas
comparacion = 'Luis, Hugo, Paco, Luis'.replace("Luis", "x").split()
original = 'Luis, Hugo, Paco, Luis'].split()

diferencias = sum([comparacion[x] != original[x] for x in range(len(original))])
print("Número de palabras que cambiaron:", diferencias)
```
