# Resumen de la Clase: Condicionales y Ciclos en Python




## 1. Estructuras Condicionales

Las estructuras condicionales nos permiten bifurcar la ejecución de nuestro código basándonos en si una condición (expresión booleana) es verdadera (True) o falsa (False).

* if: Evalúa una condición inicial.
* elif (abreviatura de else if): Evalúa condiciones adicionales si la anterior fue falsa.
* else: Ejecuta un bloque de código por defecto si ninguna de las condiciones anteriores se cumplió.


```python
# Ejemplo de Condicionales (if - elif - else)

edad = 20

if edad < 18:
    print("Eres menor de edad.")
elif edad >= 18 and edad < 65:
    print("Eres un adulto.")
else:
    print("Eres un adulto mayor.")
```


## 2. El Ciclo for

El ciclo for se utiliza para iterar sobre una secuencia (como una lista, una tupla, un diccionario, un conjunto o una cadena de texto) o utilizando la función range() para repetir una acción un número específico de veces.

```python
# Ejemplo 1: Iterar sobre una lista de frutas
frutas = ["manzana", "banana", "cereza"]
print("Iterando sobre una lista:")
for fruta in frutas:
    print(f"- Me gusta la {fruta}")

print("\nIterando usando range(start, stop):")
# Ejemplo 2: Usando range() para generar números del 1 al 5
for i in range(1, 6):
    print(f"Número: {i}")
```


## 3. El Ciclo while

El ciclo while ejecuta un bloque de código repetidamente **mientras** una condición dada se evalúe como verdadera (True). Es importante asegurarse de que la condición eventualmente sea falsa para evitar bucles infinitos.


```python
# Ejemplo de Ciclo while
contador = 1

print("Inicio del ciclo while:")
while contador <= 5:
    print(f"Contador es igual a: {contador}")
    contador += 1  # Incrementamos para evitar un ciclo infinito
print("Fin del ciclo.")
```

## 4. Control de Ciclos: break y continue

* break: Termina inmediatamente la ejecución del ciclo actual, saltando al código que sigue después del ciclo.
* continue: Salta el resto de la iteración actual y pasa directamente a la siguiente iteración del ciclo.

```python
print("Ejemplo de 'break' (detiene el ciclo al encontrar el 5):")
for numero in range(1, 10):
    if numero == 5:
        break
    print(numero)

print("\nEjemplo de 'continue' (omite los números pares):")
for numero in range(1, 6):
    if numero % 2 == 0:
        continue
    print(numero)
```
