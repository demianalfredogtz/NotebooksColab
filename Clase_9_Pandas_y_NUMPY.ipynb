### NumPy
NumPy es la biblioteca fundamental para la computación científica en Python. Facilita la creación y manipulación de vectores y matrices multidimensionales de forma eficiente.

```python
import numpy as np

# Crear un vector (array unidimensional)
vector = np.array([1, 2, 3, 4, 5])
print("Vector:", vector)

# Crear una matriz (array bidimensional)
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print("Matriz:\n", matriz)

# Operaciones matemáticas básicas elemento a elemento
vector_sumado = vector + 10
print("Vector + 10:", vector_sumado)
```

### Introducción a Pandas
Pandas es la biblioteca principal para el análisis y manipulación de datos estructurados a través de Series (unidimensionales) y DataFrames (bidimensionales, como tablas).

```python
import pandas as pd

# Crear una Serie de Pandas
serie = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("Serie de Pandas:")
display(serie)
```

### Crear y visualizar un DataFrame
Un DataFrame se puede estructurar utilizando diccionarios donde las claves son los nombres de las columnas.

```python
# Crear un DataFrame desde un diccionario
datos = {
    'Nombre': ['Ana', 'Luis', 'María'],
    'Edad': [23, 30, 25],
    'Ciudad': ['Madrid', 'Barcelona', 'Sevilla']
}
df = pd.DataFrame(datos)

print("DataFrame de ejemplo:")
display(df)
```

### Selección y filtrado básico en Pandas
Podemos seleccionar columnas específicas o filtrar filas bajo ciertas condiciones numéricas.

```python
# Seleccionar una columna
edades = df['Edad']
print("Columna Edad:")
display(edades)

# Filtrar filas (personas mayores de 24 años)
filtrado = df[df['Edad'] > 24]
print("Filtrado (Edad > 24):")
display(filtrado)
```
