```python
#Coversor Celsius a Farenheit
celsius = float(input("Ingresa los grados Celsius"))
farenheit= (celsius*9/5)+32
print(f"{celsius}ºC son {farenheit:.2f}ºF")
```

```python

# Convertidor de Kilometros a millas
km = float(input("Ingresa los kilometros a convertir"))
millas = km * 0.621371
print(f"{km} km son {millas:.2f} millas")
```

```python
# Propina de el total de la cena (10%)
cuenta = float(input("¿Cuanto fue el total de la cuenta? $"))
propina = cuenta * 0.10
total = cuenta + propina
print(f"\nPropina: ${propina:.2f}")
print(f"Total a pagar: ${total:.2f}")
```
